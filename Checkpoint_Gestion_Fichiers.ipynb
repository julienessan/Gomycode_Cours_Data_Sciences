```python
import numpy as np
```

```python
data_pret = np.genfromtxt("/content/drive/MyDrive/Colab Notebooks/Loan_prediction_dataset.csv", delimiter=",", skip_header=1, dtype=str, encoding=None)
print(data_pret)
```

```python
# Extraction des valeurs de la colonne "LoanAmount" (colonne 8) en tant que tableau numpy
col_pret = np.asarray(data_pret[:, 8], dtype=str)

# Filtrage des valeurs vides ou non numériques
col_pret_sans_valeurs_vide = (col_pret != "") & (col_pret != "nan") & (col_pret != "NaN") & (col_pret != "None") & (col_pret != "No")

# Conversion des valeurs filtrées en float
valeurs = col_pret[col_pret_sans_valeurs_vide].astype(float)

# Calcul de la moyenne, de la médiane et de l'écart type
mean_amount = np.mean(valeurs)
median_amount = np.median(valeurs)
std_amount = np.std(valeurs)
print("Moyenne des montants des prêts :", mean_amount)
print("Médiane des montants des prêts :", median_amount)
print("Écart type des montants des prêts :", std_amount)
```
