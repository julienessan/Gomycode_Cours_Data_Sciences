**Étape 2 :** Exploration des données avec ydata-profiling

```python
# Générer le rapport de profil
profile_diabete_data = ProfileReport(Diabete_data, title='Diabété Profiling Report')

# Afficher le rapport
profile_diabete_data.to_notebook_iframe()
```

**Analyser le rapport**

**1- Résumé des principales observations tirées de Pandas et du rapport de profilage de ydata:**

768 observations, 9 colonnes, toutes numériques, aucune valeur NaN déclarée. ydata ne signale aucun doublon.
Les colonnes Glucose et IMC sont presque complets (5 et 11 trous), mais l'insuline est inutilisable telle quelle, avec près d'une valeur sur deux manquante (374 valeurs manquantes).
La prévalence du diabète atteint 34,9 % (268 positifs sur 768), soit environ deux non-diabétiques pour un diabétique.

**Distributions : glucose, IMC, grossesses**


Le glucose est presque symétrique (asymétrie 0,17), centré sur 120,9 mg/dL. C'est la variable la plus complète et la mieux corrélée au diagnostic (0,49). Les diabétiques affichent une moyenne de 142 contre 111 pour les non-diabétiques, et le taux de positifs grimpe de 7 % sous 100 mg/dL à 82 % au-delà de 160.

L'IMC est légèrement étalé vers la gauche (asymétrie −0,43), moyenne à 32,0, soit déjà le seuil de l'obésité pour l'ensemble de l'échantillon. Sa corrélation avec le diagnostic est modérée (0,31). Le maximum de 67,1 est extrême mais plausible.

Les grossesses sont fortement étalées vers la droite (asymétrie 0,90), de 0 à 17, moyenne 3,8. ydata signale une corrélation forte avec l'âge, ce qui est attendu puisque le nombre de grossesses s'accumule avec le temps. Le lien direct avec le diabète reste faible (0,22), ce qui suggère que les grossesses agissent surtout comme proxy de l'âge plutôt que comme facteur propre.
