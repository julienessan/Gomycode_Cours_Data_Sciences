**Projet Python**

**objectif:**

Dans cet exercice, vous allez développer le programme de base de la calculatrice en y ajoutant des opérations mathématiques plus avancées et une gestion des erreurs en utilisant la programmation orientée objet.

```python
# Création de la classe "Calculatrice" avec Un attribut de dictionnaire pour stocker les opérations mathématiques
# Une méthode appelée "init" qui initialise le dictionnaire avec les opérations mathématiques
# Une méthode appelée "add_operation" qui prend en compte deux arguments : le symbole de l'opération et la fonction correspondante.
# Une méthode appelée "calculate" qui prend trois arguments : le premier nombre, le symbole de l'opération et le deuxième nombre.

class Calculatrice:
    def __init__(self):
        self.operations = {}

    def add_operation(self, symbol, function):
        self.operations[symbol] = function

    def calculate(self, num1, symbol, num2):
        if symbol in self.operations:
            return self.operations[symbol](num1, num2)
        else:
            raise ValueError(f"Opération '{symbol}' non reconnue.")

    def exponentiation(num1, num2):
        return num1 ** num2

    def racine_carree(num):
        if num < 0:
            raise ValueError("Impossible de calculer la racine carrée d'un nombre négatif.")
        return num ** 0.5

    def logarithme(num, base=10):
        import math
        if num <= 0:
            raise ValueError("Le logarithme n'est défini que pour les nombres positifs.")
        return math.log(num, base)
```

```python
# création d'une instance de la classe Calculator et utilisez une boucle while qui permet à l'utilisateur de continuer à effectuer des calculs jusqu'à ce qu'il décide de quitter le programme.

calc = Calculatrice()
calc.add_operation('+', lambda x, y: x + y)
calc.add_operation('-', lambda x, y: x - y)
calc.add_operation('*', lambda x, y: x * y)
calc.add_operation('/', lambda x, y: x / y if y != 0 else float('inf'))
calc.add_operation('^', Calculatrice.exponentiation)
calc.add_operation('sqrt', Calculatrice.racine_carree)
calc.add_operation('log', Calculatrice.logarithme)

while True:
    saisie_num1 = input("Entrez le premier nombre (ou 'q' pour quitter) : ").strip()
    if saisie_num1.lower() == 'q':
        break

    try:
        num1 = float(saisie_num1)
        if not isinstance(num1, (int, float)):
            raise ValueError("Le premier nombre n'est pas valide.")

        operation = input("Entrez l'opération (+, -, *, /, ^, sqrt, log) : ").strip()
        if operation == 'sqrt':
            result = calc.calculate(num1, operation, None)
        elif operation == 'log':
            saisie_base = input("Entrez la base du logarithme (par défaut 10) : ").strip()
            base = float(saisie_base or 10)
            if not isinstance(base, (int, float)):
                raise ValueError("La base n'est pas valide.")
            result = calc.calculate(num1, operation, base)
        else:
            saisie_num2 = input("Entrez le deuxième nombre : ").strip()
            num2 = float(saisie_num2)
            if not isinstance(num2, (int, float)):
                raise ValueError("Le deuxième nombre n'est pas valide.")
            result = calc.calculate(num1, operation, num2)

        print(f"Résultat : {result}")

    except ValueError as erreur:
        print(f"Erreur : {erreur}")

    cont = input("Voulez-vous continuer ? (o/n) : ").strip()
    if cont.lower() != 'o':
        break
```
