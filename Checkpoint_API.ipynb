Partie 2: Dataframe Asteroids - NeoW

```python
# Dataframe Asteroids - NeoWs

api_url = "https://api.nasa.gov/neo/rest/v1/feed?"
api_key = "61pkPtUw04gx5cCe0zFh2WFRzi39n6lCy0HG0WZB"
reponse = requests.get(api_url, params={"api_key": api_key})
```

```python
if reponse.status_code == 200 :
  results = reponse.json()
else:
  print("Error :", reponse.status_code)
```

```python
# Liste pour stocker les résultats
Liste_atro = []

# Parcourir toutes les dates et tous les astéroïdes
for date, asteroids in results["near_earth_objects"].items():
  for asteroid in asteroids:
    detail_astro = {
        "Id_asteroide" : int(asteroid["id"]),
        "Nom_asteroide" : asteroid["name"],
        "Diametre_min_Km" : float(asteroid["estimated_diameter"]["kilometers"]["estimated_diameter_min"]),
        "Magnitude_absolue" : float(asteroid["absolute_magnitude_h"]),
        "Vitesse_relative_km/s" : float(asteroid["close_approach_data"][0]["relative_velocity"]["kilometers_per_second"])
        }
    Liste_atro.append(detail_astro)

# Créer un dataframe
asteroide_df = pd.DataFrame(Liste_atro)
print(asteroide_df.head())
```

```python
# Sauvegarder en CSV
asteroide_df.to_csv("Asteroids.csv", index=False)
```
